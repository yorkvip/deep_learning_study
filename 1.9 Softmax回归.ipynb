回归 VS 分类
回归估计一个连续值
分类预测一个离散类别
## Softmax 回归输出匹配概率

- Softmax 回归用于多分类，输出每个类别的概率。
- 模型先输出未归一化分数（logits），再通过 softmax 函数转换为概率。
- Softmax 公式：$\hat{y}_i = \frac{\exp(o_i)}{\sum_k \exp(o_k)}$
- 输出满足：每个分量在 (0,1) 内，所有分量之和为 1。
- 预测类别：选择概率最大的类别。
- 训练时搭配交叉熵损失，使正确类别概率趋近 1，其余趋近 0。

## 交叉熵损失与梯度

**交叉熵损失**  
- 用于多分类任务，衡量预测概率分布与真实分布之间的差异。  
- 公式：$L = -\sum_{i=1}^K y_i \log(\hat{y}_i)$  
- 真实标签为 one-hot 时，损失简化为 $-\log(\hat{y}_{\text{正确类别}})$。  
- 损失越小，说明正确类别的预测概率越高。

**Softmax 回归中的梯度**  
- 损失对 logits 的梯度为：$\frac{\partial L}{\partial o_i} = \hat{y}_i - y_i$  
- 即预测概率减去真实 one-hot 标签。  
- 例：预测 `[0.2,0.7,0.1]`，真实 `[0,1,0]`，则梯度为 `[0.2,-0.3,0.1]`。  
- 该梯度直观地推动正确类别概率趋近 1，错误类别概率趋近 0。

**关键点**  
- Softmax 输出与交叉熵损失配合，梯度形式简洁，训练稳定高效。  
- 回归问题常用平方损失，分类问题常用交叉熵损失。

## Logits

- 定义：模型最后一层输出的未归一化分数。
- 作用：表示模型对各类别的初步倾向，分数越高越倾向该类别。
- 特点：
  - 可以是任意实数（正、负、零）。
  - 不满足概率性质：可能大于1，总和不为1。
- 经过 softmax 后转换为概率分布，才可解释为各类别的概率。
- 训练时，交叉熵损失常直接接收 logits，内部完成 softmax + 损失计算。
- 损失对 logits 的梯度 = 预测概率 - 真实标签（简洁形式）。
