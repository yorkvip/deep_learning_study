# 梯度消失与梯度爆炸

## 1. 是什么
反向传播时，浅层拿到的梯度要么几乎为 0（消失），要么大到溢出（爆炸），导致训练失败。

## 2. 为什么（根源：链式法则连乘）
某层梯度 ≈ 后面每层「权重 × 激活导数」连乘：
∂L/∂h ≈ (W₁f')·(W₂f')·…
- 连乘项数 ≈ 网络深度，越深越容易指数级变化。
- 每项 <1 → 乘积→0：**梯度消失**。
- 每项 >1 → 乘积→∞：**梯度爆炸**。

## 3. 各自表现
- 消失：浅层梯度≈0，不更新、学不到；经典元凶 sigmoid/tanh（导数最大 0.25/1，易饱和）。
- 爆炸：梯度 NaN/Inf，loss 突变、权重发散；常见于 RNN 或初始化过大。

## 4. 直观理解
- 消失：长队传话越传越轻，队首听不见 → 浅层"听不到"误差。
- 爆炸：越传越大，队首被震聋 → 训练崩。

## 5. 解决办法
- 激活换 ReLU 族（正区间导数=1，缓解消失）。
- 合理初始化：Xavier（tanh/sigmoid）、Kaiming（ReLU）。
- BatchNorm：限制激活/梯度范围。
- 残差连接（ResNet）：梯度直通浅层。
- 梯度裁剪：给梯度范数设上限，专治爆炸（RNN 常用）。
- 调小学习率。

## 6. 核心结论
- 两者都源于深层连乘：消失=乘<1 收敛到0，爆炸=乘>1 发散。
- sigmoid/tanh 易致消失；初始化过大/RNN 易致爆炸。
- ReLU + 好初始化 + BatchNorm/残差 + 梯度裁剪 是主要对策。
