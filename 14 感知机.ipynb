# 感知机（Perceptron）公式

### 1. 模型

感知机是一个二分类线性模型，输入向量 $\mathbf{x}$，输出为 $+1$ 或 $-1$：

$$
f(\mathbf{x}) = \operatorname{sign}(\mathbf{w} \cdot \mathbf{x} + b)
$$

其中：

- $\mathbf{x} = (x_1, x_2, \dots, x_n)^\top$ 为输入特征向量
- $\mathbf{w} = (w_1, w_2, \dots, w_n)^\top$ 为权重向量
- $b$ 为偏置（bias）
- $\operatorname{sign}(\cdot)$ 为符号函数（阶跃激活函数）：

$$
\operatorname{sign}(z) = \begin{cases}
+1, & z \ge 0 \\
-1, & z < 0
\end{cases}
$$

### 2. 损失函数

感知机用**误分类点到超平面的总距离**作为损失（等价于误分类点个数）：

$$
L(\mathbf{w}, b) = -\sum_{\mathbf{x}_i \in M} y_i (\mathbf{w} \cdot \mathbf{x}_i + b)
$$

其中 $M$ 为误分类点的集合，$y_i \in \{+1, -1\}$ 为真实标签。

### 3. 学习算法（随机梯度下降）

对每个误分类点 $(\mathbf{x}_i, y_i)$，沿损失下降方向更新参数：

$$
\mathbf{w} \leftarrow \mathbf{w} + \eta\, y_i\, \mathbf{x}_i
$$

$$
b \leftarrow b + \eta\, y_i
$$

其中 $\eta$ 为学习率（$0 < \eta \le 1$）。

**算法流程：**

1. 初始化 $\mathbf{w} = \mathbf{0}$，$b = 0$
2. 遍历样本，若 $y_i(\mathbf{w} \cdot \mathbf{x}_i + b) \le 0$（误分类），则更新 $\mathbf{w}$ 和 $b$
3. 重复直到没有误分类点

### 4. 收敛性（Novikoff 定理）

若训练集**线性可分**，则感知机学习算法在有限步内收敛；若线性不可分，算法不收敛（会震荡）。

## 经典激活函数

### 1. Sigmoid
$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$
- 输出范围：(0, 1)，平滑单调
- 优点：输出可解释为概率，常用于二分类输出层
- 缺点：饱和时梯度接近0（梯度消失），输出非零均值

### 2. Tanh（双曲正切）
$$
\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
$$
- 输出范围：(-1, 1)，零均值
- 优点：比 Sigmoid 梯度更强，适合隐藏层
- 缺点：仍存在梯度饱和问题

### 3. ReLU（Rectified Linear Unit）
$$
\text{ReLU}(x) = \max(0, x)
$$
- 输出范围：[0, +∞)
- 优点：计算快，缓解梯度消失，收敛更快，是隐藏层默认首选
- 缺点：负输入梯度为0，可能导致神经元“死亡”（Dead ReLU）

### 4. Leaky ReLU / PReLU
$$
\text{Leaky ReLU}(x) = \max(\alpha x, x), \quad \alpha \text{ 较小（如 }0.01\text{）}
$$
- 输出范围：(-∞, +∞)
- 优点：解决 ReLU 死亡问题，负区间保留小梯度
- 缺点：增加超参数 $\alpha$（PReLU 可学习）

### 5. Softmax
$$
\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_{j=1}^{K} e^{x_j}}
$$
- 输出范围：(0, 1)，所有分量之和为 1
- 用途：多分类输出层，将 logits 转为概率分布
- 一般与交叉熵损失配合使用

### 激活函数选择建议
- 隐藏层优先使用 ReLU；若担心死亡神经元，可尝试 Leaky ReLU 或 ELU。
- 二分类输出层：Sigmoid + 二元交叉熵。
- 多分类输出层：Softmax + 交叉熵。
- 回归输出层：通常不用激活函数（线性输出）。

## 多层感知机（MLP）总结笔记

### 1. 定义
多层感知机是一种前馈神经网络，由输入层、一个或多个隐藏层和输出层组成。它通过隐藏层和非线性激活函数，能够拟合复杂的非线性映射，是深度学习的基础模型。

### 2. 动机
- 线性模型（如线性回归、softmax回归）只能学习线性关系。
- 现实问题多为非线性，需要引入隐藏层和激活函数增强模型表达能力。

### 3. 结构
- **输入层**：接收特征向量，特征数为 d。
- **隐藏层**：一个或多个全连接层，每层后接非线性激活函数。
- **输出层**：根据任务输出结果，回归通常线性输出，分类常用 softmax。

### 4. 前向传播（以单隐藏层为例）
设输入 $X \in \mathbb{R}^{n \times d}$，隐藏层神经元数为 h，输出维度为 q。

$$
H = \sigma(XW_1 + b_1)
$$
$$
O = H W_2 + b_2
$$

其中：
- $W_1$：形状 $d \times h$，$b_1$：形状 $h$
- $W_2$：形状 $h \times q$，$b_2$：形状 $q$
- $\sigma$：激活函数

### 5. 激活函数
- **ReLU**：$\max(0,x)$，隐藏层首选，计算快，缓解梯度消失。
- **Sigmoid**：输出(0,1)，适合二分类输出层，但易饱和。
- **Tanh**：输出(-1,1)，零均值，比 Sigmoid 好，但仍有饱和问题。
- **Softmax**：多分类输出层，将 logits 转为概率分布。

### 6. 训练过程
1. 前向传播计算预测输出
2. 用损失函数计算误差（回归：均方损失；分类：交叉熵）
3. 反向传播求各参数梯度
4. 优化器更新参数（如 SGD、Adam）

### 7. 参数数量
以单隐藏层为例：
$$
d \times h + h + h \times q + q
$$
包括输入到隐藏层权重和偏置、隐藏层到输出层权重和偏置。

### 8. 作用与意义
- 万能逼近：在足够多神经元下，可近似任意连续函数。
- 是更复杂网络（CNN、RNN、Transformer）的基础模块。
- 通过堆叠多层形成深度网络，可学习层次化特征。

### 9. 关键点
- 没有激活函数，多层线性叠加仍是线性。
- 隐藏层神经元数量影响模型容量，过多易过拟合，过少欠拟合。
- 实践中优先使用 ReLU，并根据任务选择输出层激活。
