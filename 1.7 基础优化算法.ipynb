## 梯度下降更新公式

$$
\theta_{t+1} = \theta_t - \eta \nabla_\theta J(\theta_t)
$$

### 符号解释

- $\theta$：模型参数（向量）
- $t$：当前迭代步数
- $\eta$：学习率（步长，控制每次更新的幅度）
- $J(\theta)$：损失函数（目标函数，衡量模型预测与真实值的差距）
- $\nabla_\theta J(\theta_t)$：损失函数 $J$ 关于参数 $\theta$ 在 $\theta_t$ 处的梯度（指向损失函数上升最快的方向）
- 更新方向：$-\nabla_\theta J(\theta_t)$ 表示沿梯度反方向（即损失函数下降最快的方向）更新参数


相当于每次往函数的导数的零点--损失函数下降最快的方向走一小步-逐步逼近

## 小批量随机梯度下降（Mini-batch SGD）

**更新公式：**
$$
\theta_{t+1} = \theta_t - \eta \frac{1}{|B|} \sum_{i \in B} \nabla_\theta J(\theta_t; x^{(i)}, y^{(i)})
$$

- $B$：一个小批量样本索引集合
- $|B|$：批量大小（batch size）
- $\eta$：学习率

**关键点：**
- 每次迭代随机抽取一个小批量样本，计算梯度并更新。
- 权衡：批量大 → 梯度估计准，但计算慢；批量小 → 计算快，但梯度噪声大。
- 常用批量大小：32、64、128 等。
- 比批量GD更高效，比随机GD更稳定。

两个重要的参数是批量大小和学习率
