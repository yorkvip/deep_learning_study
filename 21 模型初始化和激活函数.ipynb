# 为什么每层方差要是常数？

## 动机：防止梯度消失 / 梯度爆炸

深层网络是一条长链：梯度要一路乘回前面每一层的权重，梯度方差约等于 $D^l$（每层方差 $D$ 的幂次）。

| 情况 | 结果 |
| --- | --- |
| $D < 1$ | $D^l \to 0$ → **梯度消失**，前面层学不进去 |
| $D > 1$ | $D^l \to \infty$ → **梯度爆炸**，数值变 `nan`，训练崩溃 |

让 $D \approx 1$（各层方差恒定），信号在任意深度既不衰减也不放大 → 同时避免两个极端。

## 体现在两个方向

- **前向**：输出标准差稳定，激活值不饱和
- **反向**：梯度尺度一致（更关键）

## 为什么是"方差"不是"期望"

消失/爆炸的本质是**波动（二阶统计量）的连乘累积**，不是均值漂移。

## 由此诞生的初始化方法

- **Xavier**：用于线性 / 对称激活（如 sigmoid、tanh）
- **He**：用于 ReLU（方差加倍）

两者的推导前提正是"每层方差不变"。
