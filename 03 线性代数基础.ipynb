你可以将向量视作标量值组成的列表

```python
import torch
x=torch.arange(4)
print(x)
```

通过张量的索引来访问任意元素

```python
print(x[3])
```

访问张量的长度

```python
print(len(x))
```

通过指定m,n来创建形状为m X n的矩阵

```python
A=torch.arange(20).reshape(5,4)
print(A)
print(A.T)
```

两个矩阵按元素的乘法，在数学上称为“哈达玛积”

```python
B=A.clone()
print(A*B)
```

计算元素和

```python
print(A.sum())
print(A.shape) # 这个不要 shape 不要加括号 ~
```

指定求和汇总张量的轴

```python
A_sum_axis0=A.sum(axis=0) # 求和每一列
print(A_sum_axis0)
print(A_sum_axis0.shape)
```

```python
A_sum_axis1=A.sum(axis=1) # 计算每一行的和
print(A_sum_axis1)
print(A_sum_axis1.shape)
```

```python
print(A.sum(axis=[0,1]).shape)
```

求平均值

```python
# A 是整数张量，mean() 只支持浮点/复数类型，用 .float() 转成浮点再求均值
print(A.float().mean(axis=(0,1)))
```

计算总和或均值时，保持轴数不变

```python
sum_A=A.sum(axis=1,keepdims=True)
print(sum_A)
```

可以直接广播将 A 除以 sum_A

```python
print(A/sum_A)
```

某个轴计算A元素的累加求和

```python
print(A.cumsum(axis=0))
```

点积是相同位置的按元素乘积的和

```python
y=torch.ones(4,dtype=torch.float32) # 全是1元素
x=x.float()
print(torch.dot(x,y)) # 是一个标量
'''
相当于
print(torch.sum(x * y))
'''
```

矩阵向量积 Ax 是一个长度为 m 的向量。相当于点积 a.T * X

```python
# A 是 int64(Long) 类型，x 也要转成同类型（long），torch.mv 要求两个张量 dtype 一致
x=x.long()
print(torch.mv(A,x))
```

矩阵乘法！

```python
B=torch.ones(4,3,dtype=torch.int64)
print(torch.mm(A,B))
```

L2范数是向量元素平方和的平方根

```python
u=torch.tensor([3.0,-4.0])
print(torch.norm(u))
```

L1范数，表示为向量元素的绝对值求和

```python
print(torch.abs(u).sum())
```

F范数：矩阵元素平方和的平方根,相当于是L2范数的扩展

```python
print(torch.norm(torch.ones(4,9)))
```
