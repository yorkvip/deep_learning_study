1.1 一维对象的创建

```python
#创建字典
dict_v={'a':0,"b":0.25,'c':0.5,'d':0.75,'e':1}

#字典创建对象
sr=pd.Series(dict_v)
print(sr)
```

```python
#数组创建对象
v=[0,1,2,3,4] #这个改成数组也可以，张量也可以
k=['a','b','c','d','e']
sr=pd.Series(v,index=k)
print(sr)
```

Series有两个属性

index

values

```python
print(sr.values) #无论sr是由什么传入的，都是以数组存储的
print(sr.index)
```

Series的延展方向是竖直方向，DataFrame延展方向是水平方向

每个Series就是一列数据，相当于对一列一列的数据做拼接

```python
a=['1号','2号','3号','4号']
b=['16','18','19','20']
c=['女','男','男','男']
sr1=pd.Series(b,index=a)
sr2=pd.Series(c,index=a)
df=pd.DataFrame({'年龄':sr1,'性别':sr2})
print(df)
```

如果sr1和sr2的行标签不完全一致，会取交集，有NaN

```python
c=['年龄','性别']
i=['一号','二号','三号','四号']
v=[[53,'女'],[64,'男'],[72,'男'],[82,'女']]
df=pd.DataFrame(v,index=i,columns=c)
print(df)
```

DataFrame对象有三个属性：values,index,columns

```python
print(df.values)
print(df.index)
print(df.columns)
# 提取某一列
arr=df.values[:,0]
print(arr)
```

## 二维对象的索引

二维对象中。索引器不能去掉，否则会报错，必须适应索引器的存在

```python
a=['1号','2号','3号','4号']
b=['16','18','19','20']
c=['女','男','男','男']
sr1=pd.Series(b,index=a)
sr2=pd.Series(c,index=a)
df=pd.DataFrame({'年龄':sr1,'性别':sr2})
```

```python
print(df.loc['1号','年龄'])
# 花式索引
print(df.loc[['1号','3号'],['性别','年龄']])
```

## 访问切片！

```python
c=['年龄','性别']
i=['一号','二号','三号','四号']
v=[[53,'女'],[64,'男'],[72,'男'],[82,'女']]
df=pd.DataFrame(v,index=i,columns=c)
print(df)
```

```python
print(df.loc['一号':'三号','年龄'])
```

## 对象的变形

```python
df=df.T
print(df)
df=df.T
#当行列搞反了
```

```python
print(df)

# 左右反转
df=df.iloc[:,::-1]
print(df)

# 上下反转
df=df.iloc[::-1,:]
print(df)
```

## 对象的重塑

```python
a=['1号','2号','3号','4号']
b=['16','18','19','20']
c=['女','男','男','男']
d=[1,2,3,4]
sr1=pd.Series(b,index=a)
sr2=pd.Series(c,index=a)
df=pd.DataFrame({'年龄':sr1,'性别':sr2})
print(df)
```

```python
sr3=pd.Series(d,index=a)
df['牌照']=sr3
print(df)
```

对象和对象之间的运算

```python
sr1=pd.Series([1,2,3],index=['1号','2号','3号'])
sr2=pd.Series([1,2,3,4],index=['1号','2号','3号','4号'])
print(sr1+sr2)
```

```python
a=['1号','2号','3号','4号']
b=['16','18','19','20']
c=['女','男','男','男']
d=[1,2,3,4]
sr1=pd.Series(b,index=a)
sr2=pd.Series(c,index=a)
df=pd.DataFrame({'年龄':sr1,'性别':sr2})
sr3=pd.Series(d,index=a)
df['牌照']=sr3
print(df)
```

```python
df['加法']=df['年龄'].astype(int)+df['牌照']
print(df)
```

## 剔除缺失值

主要用  .dropna()方法

```python
# 创建 sr
v=[53,None,72,82]
k=['1号','2号','3号','4号']
sr=pd.Series(v,index=k)
print(sr)
```

```python
print(sr.dropna())
```

对于二维数据，如果一个数据是NaN就全部剔除，太过残忍。可以设置如下：

df.dropna(how='all')

只有该行全是NaN才会被剔除

all 可以改成数字-->这一行超过 某个数字后，才可以剔除这一行

```python
print(sr.fillna(0))
```

```python
import numpy as np
print(sr.fillna(np.mean(sr)))
```

```python
# 用前值或者后值填充
print(sr.fillna(method='ffill'))
print(sr.fillna(method='bfill'))
```
