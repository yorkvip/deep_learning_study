## 均方损失（Mean Squared Error, MSE）

**公式：**
$$
L = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

- $n$：样本数量
- $y_i$：第 $i$ 个样本的真实值
- $\hat{y}_i$：第 $i$ 个样本的预测值

**用途**  
常用于回归问题，衡量预测值与真实值之间的平均平方误差。

## 绝对值损失（L1 Loss / Mean Absolute Error, MAE）

**单个样本损失：**
$$
L = |y - \hat{y}|
$$

**整体平均损失：**
$$
L = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

- $n$：样本数量
- $y_i$：第 $i$ 个样本的真实值
- $\hat{y}_i$：第 $i$ 个样本的预测值

**用途**  
常用于回归问题，对异常值（离群点）比均方损失更鲁棒，因为误差不会被平方放大。

## Huber 损失（结合 MSE 与 MAE）

**单个样本损失：**
$$
L_\delta(y, \hat{y}) =
\begin{cases}
\frac{1}{2}(y - \hat{y})^2 & \text{若 } |y - \hat{y}| \le \delta \\
\delta |y - \hat{y}| - \frac{1}{2}\delta^2 & \text{若 } |y - \hat{y}| > \delta
\end{cases}
$$

- $y$：真实值
- $\hat{y}$：预测值
- $\delta$：阈值超参数，决定何时从平方损失切换为线性损失

**整体平均损失：**
$$
L = \frac{1}{n} \sum_{i=1}^{n} L_\delta(y_i, \hat{y}_i)
$$

**特点**  
- 小误差：用平方损失，梯度随误差减小而变小，收敛精细。  
- 大误差：用线性损失，梯度恒定，对离群点不敏感，比 MSE 更鲁棒。  
- 常用在回归任务中，尤其是数据可能含有异常值时。
